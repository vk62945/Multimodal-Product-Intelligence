15. Remove incomplete multimodal records

```python
df_model = df_clean[
    df_clean["subCategory"].isin(TARGET_CLASSES)
].copy()
print("Before removing incomplete records:", len(df_model))
df_model = df_model.dropna(
    subset=["productDisplayName"]
).copy()

df_model = df_model[
    df_model["id"].isin(image_ids)
].copy()
print("After removing incomplete records:", len(df_model))
print("Missing text:", df_model["productDisplayName"].isnull().sum())
print("Missing images:", (~df_model["id"].isin(image_ids)).sum())
```

16. Create Image Path

```python
IMAGE_DIR = DATA_DIR / "images"

df_model["image_path"] = df_model["id"].apply(
    lambda x: str(IMAGE_DIR / f"{int(x)}.jpg")
)
```

17. Create numeric labels

```python
from sklearn.preprocessing import LabelEncoder
label_encoder = LabelEncoder()
df_model["label"] = label_encoder.fit_transform(df_model["subCategory"])
```

```python
label_mapping = dict(
    zip(
        label_encoder.classes_,
        label_encoder.transform(label_encoder.classes_)
    )
)
label_mapping = {
    label: int(index)
    for label, index in label_mapping.items()
}
label_mapping
```

18. Save the class mapping

```python
import json

with open("../data/processed/label_mapping.json", "w") as file:
    json.dump(label_mapping, file, indent=4)
```

21. Check duplicate product names

```python
duplicate_test_count = (
    df_final["productDisplayName"]
    .duplicated()
    .sum()
)
print("Duplicate product names:", duplicate_test_count)
```

```python
unique_text_count = (
    df_final["productDisplayName"]
    .nunique()
)
print("Unique product names:", unique_text_count)
```

```python
text_label_groups = (
    df_final
    .groupby("productDisplayName")["subCategory"]
    .nunique()
)
print(
    "Product names appearing with multiple labels:",
    (text_label_groups > 1).sum()
)
```

```python
print(
    text_label_groups
    .value_counts()
    .sort_index()
)
```

22. Remove ambiguous product names

```python
df_final.shape
```

```python
ambiguous_names = text_label_groups[
    text_label_groups > 1
].index

print("Ambiguous product names:", len(ambiguous_names))

df_split = df_final[
    ~df_final["productDisplayName"].isin(ambiguous_names)
].copy()

print("Dataset after removing ambiguous names:", df_split.shape)
```

```python
check_groups = (
    df_split.groupby("productDisplayName")["subCategory"].nunique()
)
print(
    "Product names with multiple labels:",
    (check_groups > 1).sum()
)
```

23. Create grouped split

```python
from sklearn.model_selection import GroupShuffleSplit

gss = GroupShuffleSplit(
    n_splits = 1,
    test_size = 0.30,
    random_state = 42
)

train_idx, temp_idx = next(
    gss.split(df_split, groups = df_split["productDisplayName"])
)

train_df = df_split.iloc[train_idx].copy()
temp_df = df_split.iloc[temp_idx].copy()
print("Train:", train_df.shape)
print("Temp:", temp_df.shape)
```

```python
gss_test = GroupShuffleSplit(
    n_splits =1,
    test_size = 0.50,
    random_state = 42
)

val_idx, test_idx = next(
    gss_test.split(temp_df, groups = temp_df["productDisplayName"])
)

val_df = temp_df.iloc[val_idx].copy()
test_df = temp_df.iloc[test_idx].copy()

print("Train:", len(train_df))
print("Validation:", len(val_df))
print("Test:", len(test_df))
```

```python
train_names = set(train_df["productDisplayName"])
val_names = set(val_df["productDisplayName"])
test_names = set(test_df["productDisplayName"])
print("train n Validation:", len(train_names & val_names))
print("train n Test:", len(train_names & test_names))
print("Validation n Test:", len(val_names & test_names))
```

```python
print("TRAIN")
print(train_df["subCategory"].value_counts())
print("VALIDATION")
print(val_df["subCategory"].value_counts())
print("TEST")
print(test_df["subCategory"].value_counts())
```

```python
train_classes = set(train_df["subCategory"])
val_classes = set(val_df["subCategory"])
test_classes = set(test_df["subCategory"])
print("Classes missing from train:", set(TARGET_CLASSES) - train_classes)
print("Classes missing from validation:", set(TARGET_CLASSES) - val_classes)
print("Classes missing from test:", set(TARGET_CLASSES) - test_classes)
```

### Leakage-Safe Dataset Split
